```python

import tsplib95 as tsp
import numpy as np
import random

try:
  from google.colab import drive
  drive.mount('/content/drive')
  mounted=True
except ImportError:
  mounted=False
```

```python
try:
  if mounted:
    problem=tsp.load('/content/drive/MyDrive/ali535.tsp')

  else:
    problem=tsp.load('ali535.tsp')
  N=problem.dimension

  distance=np.zeros((N,N))

  for i in range(N):
    for j in range(N):
      distance[i, j]=problem.get_weight(i + 1, j + 1)
  print(f"Successfully loaded problem with {N} cities.")

except Exception as e:
  print(f"Error loading or parsing file: {e}")
```

```python
def calc_cost(tour,distance):
  cost = 0
  N = len(tour)
  for i in range(N):
    city_a = tour[i]
    city_b = tour[(i+1)%N]
    cost += distance[city_a,city_b]
  return cost
```

```python
def initial_solution(cities):
  solution = list(range(cities))
  random.shuffle(solution)
  return solution
```

```python
def generate_neighbours(tour):
  N = len(tour)
  i, j = random.sample(range(N),2)
  i, j = min(i, j),max(i, j)

  neighbour=tour[:]
  neighbour[i:j+1]=neighbour[i:j+1][::-1]
  return neighbour
```

```python
def hill_climbing(distance, max_i=50000):
  cities = distance.shape[0]
  tour = initial_solution(cities)
  cost = calc_cost(tour,distance)

  for i in range(max_i):
    neighbour = generate_neighbours(tour)
    neighbour_cost = calc_cost(neighbour,distance)

    if neighbour_cost<cost:
      tour=neighbour
      cost=neighbour_cost
  return tour,cost
```

```python
try:
   final_tour,final_cost=hill_climbing(distance)
   print(f"search complete.")
   print(f"Final cost: {final_cost: .2f}")
   print(f"Final tour of first 10 cities: {final_tour[:10]}")
   print(f"Tour length: {len(final_tour)}")
   print(f"Total cities: {distance.shape[0]}")

except Exception as e:
  print(f"Error during optimization: {e}")
```
